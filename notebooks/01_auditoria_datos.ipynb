```python
# Inspeccion base de datos SPSS

import pyreadstat
from pathlib import Path

ruta = Path("../data/raw/spss/enemdu_persona_2026_l_trimestre.sav")
print("Existe:", ruta.exists())

# metadataonly=True lee solo la estructura, sin cargar los datos

_, meta = pyreadstat.read_sav(ruta, metadataonly=True)

print("Filas:", meta.number_rows,"| Columnas:", meta.number_columns)
print(meta.column_names)
```

```python
# Buscar variables que necesitamos

nombres = [c.lower() for c in meta.column_names]

for clave in ["fexp", "condact", "ingrl", "sexo", "p02", "area", "prov", "p03"]:
    print(clave, "->", [n for n in nombres if clave in n][:6])
```

```python
# Etiquetas de variables en tabla

import pandas as pd
etiquetas = pd.DataFrame({
    "variable": meta.column_names,
    "etiqueta": [meta.column_names_to_labels[c] for c in meta.column_names],
})
etiquetas.to_csv("../docs/variables_y_etiquetas.csv", index=False, encoding="utf-8-sig")
etiquetas.head(40)
```

```python
# Busqueda variables de interes

candidatas = ["fexp", "condact", "ingrl", "p02", "p03", "area", "ciudad",
              "empleo", "secemp", "grupo1", "rama1", "nnivins", "escolaridad"]

existen = [c for c in candidatas if c in meta.column_names]
faltan = [c for c in candidatas if c not in meta.column_names]

for c in existen:
    print(f"\n{c}: {meta.column_names_to_labels.get(c)}")
    print("   códigos:", meta.variable_value_labels.get(c, "sin etiquetas de valores"))

print("\nNo existen con ese nombre:", faltan)
```

```python
# Diagnostico de provincia

import sys
sys.path.append("..")
from src.data import cargar_enemdu

df = cargar_enemdu()

# 1) longitud de los códigos de ciudad
cod = df["ciudad"].astype(int).astype(str)
print(cod.str.len().value_counts(), "\n")

# 2) prefijo de provincia, peso medio y tamaño de muestra
a_texto = lambda s: s.astype(str).str.replace(r"\.0$", "", regex=True).str.zfill(6)
df["prov_cod"] = a_texto(df["ciudad"]).str[:2]
print(df.groupby("prov_cod")["fexp"].mean().round(0).sort_values(), "\n")

# 3) dominios presentes en provincias dudosas
for p in ["14", "24", "06", "09"]:
    print(p, "-> dominios:", sorted(df.loc[df["prov_cod"] == p, "dominio"].unique()))

# 4) población y casos por dominio
print(df.groupby("dominio")["fexp"].agg(["size", "sum"]).round(0))
```
